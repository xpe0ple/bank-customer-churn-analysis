## Bivariate Analysis

```python
# Gender vs Churn
sns.countplot(
    x='gender',
    hue='churn',
    data=df
)
```

```python
gender_churn = (
    df.groupby('gender')['churn']
    .mean()
    .mul(100)
    .reset_index()
)

sns.barplot (
    data=gender_churn,
    x='gender',
    y='churn'
)

plt.ylabel('Churn Rate (%)')
plt.title('Churn rate by Gender')
plt.show()
```

```python
df['AgeGroup'] = pd.cut(
    df['age'],
    bins=[0,18,30,45,60,100],
    labels=['0-18','19-30','31-45','46-60','60+']
)

age_churn = (
    df.groupby('AgeGroup')['churn']
      .mean()
      .mul(100)
      .reset_index()
)

sns.barplot(
    data=age_churn,
    x='AgeGroup',
    y='churn'
)

plt.title('Churn Rate by Age Group')
plt.ylabel('Churn Rate (%)')
plt.show()
```

```python
occupation_churn = (
    df.groupby('occupation')['churn']
      .mean()
      .mul(100)
      .sort_values(ascending=False)
      .head(10)
      .reset_index()
)

occupation_churn
```

```python
plt.figure(figsize=(10,5))

sns.barplot(
    data=occupation_churn,
    x='occupation',
    y='churn'
)

plt.title('Top 10 Occupation by Churn Rate')
plt.ylabel('Churn Rate (%)')
plt.xlabel('Occupation')

plt.xticks(rotation=45)
plt.show()
```

```python
customer_category_churn = (
    df.groupby('customer_nw_category')['churn']
      .mean()
      .mul(100)
      .reset_index()
)

customer_category_churn
```

```python
plt.figure(figsize=(8,5))

sns.barplot(
    data=customer_category_churn,
    x='customer_nw_category',
    y='churn'
)

plt.title('Churn Rate by Customer Category')
plt.ylabel('Churn Rate (%)')
plt.xlabel('Customer Category')

plt.show()
```

## Correlation Analysis

```python
numeric_cols = [
    'age',
    'current_balance',
    'current_month_credit',
    'current_month_debit',
    'vintage',
    'churn'
]
```

```python
corr = df[numeric_cols].corr()

plt.figure(figsize=(8,6))
sns.heatmap(corr, annot=True)

plt.title('Correlation Matrix')
plt.show()
```
